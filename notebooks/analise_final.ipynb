## 3. Visão geral do negócio

```python
def calcular_kpis(dados):
    faturamento=dados['Sales'].sum()
    lucrototal=dados['Profit'].sum()
    margemglobal=lucrototal/faturamento
    quantidadevendida=dados['Quantity'].sum()
    totalpedidos=dados['Order ID'].nunique()
    ticketmedio=faturamento/totalpedidos

    kpis={
    'faturamento_total': faturamento,
    'lucro_total': lucrototal,
    'margem_global': margemglobal,
    'quantidade_vendida': quantidadevendida,
    'total_pedidos': totalpedidos,
    'ticket_medio': ticketmedio
    }

    return kpis
    

```

```python
kpis=calcular_kpis(df)

print(f"Faturamento total: US$ {kpis['faturamento_total']:.2f}")
print(f"Lucro total: US$ {kpis['lucro_total']:.2f}")
print(f"Margem global: {kpis['margem_global']:.2%}")
print(f"Quantidade vendida: {kpis['quantidade_vendida']:}")
print(f"Total de pedidos: {kpis['total_pedidos']:}")
print(f"Ticket médio: US$ {kpis['ticket_medio']:.2f}")
```

### Interpretação dos KPIs

A empresa apresentou lucro no período analisado. O faturamento foi de aproximadamente US$ 2.30 milhões, com lucro total de US$ 286,40 mil.

A margem global foi de 12,47%. Isso significa que, para cada US$ 100 faturados, aproximadamente US$ 12,47 permaneceram como lucro.

O ticket médio foi de US$ 458,61. Isso representa o valor médio faturado por pedido, considerando os 5.009 pedidos únicos registrados na base.

Apesar do resultado geral positivo, ainda é necessário investigar se existem categorias, regiões ou períodos com baixa lucratividade ou prejuízo assim como foi feito na exploração inicial onde achamos categorias com prejuizo.

## 4. Análise temporal

### 4.1 Desempenho anual

```python
def resumir_por_grupo(dados, coluna_grupo):
    resumo=dados.groupby(coluna_grupo).agg(
        faturamento_total=('Sales','sum'),
        lucro_total=('Profit','sum'),
        quantidade_vendida=('Quantity','sum'),
        total_pedidos=('Order ID','nunique'),
    )
    resumo['margem_global']=(
        resumo['lucro_total']/resumo['faturamento_total']
    )
    resumo['ticket_medio']=(
        resumo['faturamento_total']/resumo['total_pedidos']
    )
    return resumo

def variacoes_temporais(resumo):
    resumo_temporal=resumo.copy()
    resumo_temporal['variacao_faturamento']=(
        resumo_temporal['faturamento_total'].pct_change()
    )
    resumo_temporal['variacao_lucro']=(
        resumo_temporal['lucro_total'].pct_change()
    )
    return resumo_temporal

resumo_anual_base=resumir_por_grupo(df, 'ano')
resumo_anual=variacoes_temporais(resumo_anual_base)
resumo_anual
```

### Interpretação do resumo anual

O ano de 2016 apresentou a maior margem global, com 13,43%, enquanto 2014 teve a menor margem, com 10,23%.

Em 2015, o faturamento foi menor do que em 2014. Mesmo assim, o lucro aumentou e a margem passou de 10,23% para 13,10%. Isso indica uma melhora na eficiência financeira das vendas realizadas naquele ano.

Ainda é necessário investigar quais fatores contribuíram para essa melhora, como descontos, categorias vendidas ou desempenho regional.

A quantidade vendida aumentou em todos os anos. Em 2015, foram vendidas mais unidades do que em 2014, mas o faturamento foi o menor de todo o período analisado.

Isso indica que o aumento no volume de produtos vendidos não resultou em crescimento do faturamento. Uma possível explicação é a venda de produtos com menor valor médio ou uma mudança no comportamento dos descontos, hipóteses que ainda precisam ser investigadas.

O ticket médio caiu de US$ 499,74 em 2014 para US$ 453,31 em 2015. Isso reforça a conclusão de que, apesar do aumento no número de pedidos e na quantidade vendida, cada pedido gerou menos faturamento em média.

Em 2017, o ticket médio foi ainda menor, com US$ 434,63. Mesmo assim, o ano apresentou o maior faturamento do período, sustentado pelo aumento expressivo no número de pedidos e na quantidade vendida.

Esses resultados mostram que o crescimento do faturamento pode vir do aumento do volume de pedidos, mesmo quando o valor médio de cada pedido diminui.

variação faturamento:

variação de lucro:

Em 2015, o faturamento caiu 2,83%, mas o lucro cresceu 24,37%. Esse contraste reforça que houve melhora na eficiência financeira, embora a causa ainda precise ser investigada.

Em 2016, o lucro cresceu 32,74%, superando o crescimento de 29,47% do faturamento. Como o lucro avançou mais rapidamente, a margem global aumentou.

Em 2017, o faturamento cresceu 20,36%, enquanto o lucro cresceu apenas 14,24%. Como o lucro avançou mais lentamente que o faturamento, a margem global diminuiu.

```python
formatacao_resumo={
    'faturamento_total': 'US$ {:,.2f}',
    'lucro_total': 'US$ {:,.2f}',
    'quantidade_vendida': '{:,.0f}',
    'total_pedidos': '{:,.0f}',
    'margem_global': '{:.2%}',
    'ticket_medio': 'US$ {:,.2f}',
    'variacao_faturamento': '{:.2%}',
    'variacao_lucro': '{:.2%}'
}


resumo_anual_formatado=resumo_anual.style.format(
    formatacao_resumo,
    na_rep='-',
    decimal=',',
    thousands='.'
)


resumo_anual_formatado
```

### Interpretação das faixas de desconto

As vendas com desconto agressivo apresentaram faturamento menor e menor volume, mas geraram lucro e margem global negativos nos dois anos. Isso mostra que, apesar de ocorrerem com menor frequência, tiveram impacto financeiro relevante.

Os descontos altos, por outro lado, apresentaram lucro e margem positivos. Em 2015, essa faixa aumentou o lucro e melhorou a margem global, indicando que nem todo desconto foi prejudicial.

A melhora de Technology em 2015 parece estar associada ao crescimento das vendas sem desconto e à maior eficiência das vendas com desconto alto. Os descontos agressivos continuaram sendo um grupo problemático.

```python
resumo_2014=resumo_faixa_desconto.loc[2014]
resumo_2015=resumo_faixa_desconto.loc[2015]
diferencas_desconto=pd.DataFrame({
    'diferenca_faturamento':(
        resumo_2015['faturamento_total']-resumo_2014['faturamento_total']
    ),

    'diferenca_lucro':(
        resumo_2015['lucro_total']-resumo_2014['lucro_total']
    ),

    'aumento_margem_pontos_percentuais':(
        resumo_2015['margem_global']-resumo_2014['margem_global']
    )*100,

    'diferenca_pedidos':(
        resumo_2015['total_pedidos']-resumo_2014['total_pedidos']
    )
})

formatacao_diferencas={
    'diferenca_faturamento': 'US$ {:,.2f}',
    'diferenca_lucro': 'US$ {:,.2f}',
    'aumento_margem_pontos_percentuais':('{:,.2f} pontos percentuais'),
    'diferenca_pedidos': '{:,.0f}',
}

diferencas_formatado=diferencas_desconto.style.format(
    formatacao_diferencas,
    na_rep='-',
    decimal=',',
    thousands='.'
)

diferencas_formatado
```

### Interpretação das diferenças de desconto

A melhora do lucro de Technology em 2015 foi impulsionada principalmente pelo crescimento das vendas sem desconto, que apresentaram o maior aumento de lucro.

As vendas com desconto alto também contribuíram, pois aumentaram o lucro e melhoraram a margem mesmo com menor faturamento.

Os descontos agressivos continuaram sendo prejudiciais. Apesar da redução do volume financeiro dessa faixa, ela permaneceu com lucro e margem negativos.

Portanto, a melhora de Technology está associada ao crescimento das vendas sem desconto e à maior eficiência dos descontos altos, e não a uma melhora das vendas com descontos agressivos.
