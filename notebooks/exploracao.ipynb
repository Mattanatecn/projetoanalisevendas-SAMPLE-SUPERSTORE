```python
# Importa a biblioteca Pandas, essencial para manipulação e análise de tabelas de dados
import pandas as pd
```

```python
# Carrega o conjunto de dados bruto. 
# encoding='latin1' é usado porque o arquivo possui caracteres especiais (acentos, etc) que o padrão UTF-8 não reconhece
df = pd.read_csv('../dados/bruto/Sample - Superstore.csv', encoding='latin1')
```

```python
# Exibe as primeiras 5 linhas do DataFrame para ter uma visão geral das colunas e dos dados
df.head()
```

```python
# Analisa a estrutura do DataFrame: tipos de dados, se há valores nulos e o volume de memória usado
df.info()
```

```python
# Verifica a existência de linhas idênticas para garantir que não haja contagem duplicada de vendas
print(f"Linhas duplicadas: {df.duplicated().sum()}")
```

```python
# Cálculo do KPI de Vendas Totais: soma de todo o faturamento da empresa
vendas=df['Sales'].sum()
print(vendas)
```

```python
# Cálculo do KPI de Lucro Total: soma de todos os lucros (positivos e negativos)
lucro_total=df['Profit'].sum()
print(lucro_total)
```

```python
# Identifica as vendas que geraram prejuízo (lucro < 0) e soma esse prejuízo total
# Isso ajuda a entender a escala do problema financeiro da empresa
vendas_prejuizo=df[df['Profit'] < 0]
lucro_negativo=vendas_prejuizo['Profit'].sum()
print(lucro_negativo)
```

```python
# Agrupa os dados por Categoria para descobrir qual setor do negócio é mais lucrativo
lucro_categoria=df.groupby('Category')['Profit'].sum()
print(lucro_categoria)
```

```python
# Detalha o lucro por Subcategoria para localizar exatamente onde o dinheiro está sendo perdido
lucro_subcategorias=df.groupby('Sub-Category')['Profit'].sum()
print(lucro_subcategorias)
```

```python
# Ordena os lucros das subcategorias do menor para o maior
# Isso evidencia rapidamente a subcategoria 'Tables' como a principal fonte de prejuízo
lucro_ordenado=lucro_subcategorias.sort_values()
print(lucro_ordenado)
```

```python
# PASSO 4: INVESTIGAÇÃO DA CAUSA RAIZ
# Isolamos apenas as vendas de 'Tables' para analisar o comportamento específico desta subcategoria
tables=df[df['Sub-Category']=='Tables']
print(tables)
```

```python
# Filtrando vendas de mesas com descontos agressivos (20% ou mais)
# Hipótese: Descontos altos estão anulando a margem de lucro
df_high_discount=tables[tables['Discount']>=0.2]
print(df_high_discount)
```

```python
# Filtrando vendas de mesas com descontos baixos (menos de 20%)
# Serve como grupo de controle para comparar se o prejuízo persiste sem descontos altos
df_low_discount=tables[tables['Discount']<0.2]
print(df_low_discount)
```

```python
# Calcula o lucro total para o grupo de descontos altos
# Resultado esperado: Lucro fortemente negativo
lucro_negativo=df_high_discount['Profit'].sum()
print(f"Lucro Total (Descontos >= 20%): {lucro_negativo}")
```

```python
# Calcula o lucro total para o grupo de descontos baixos
# Resultado esperado: Lucro positivo ou prejuízo irrelevante
lucro_positivo=df_low_discount['Profit'].sum()
print(f"Lucro Total (Descontos < 20%): {lucro_positivo}")
```

```python
# LIMPEZA DE DADOS: Transformando colunas de data de texto (str) para formato datetime
# Isso permite realizar cálculos de tempo, sazonalidade e análises temporais
df['Order Date']=pd.to_datetime(df['Order Date'])
df['Ship Date']=pd.to_datetime(df['Ship Date'])
df.info()
```

```python
dado_nulo=df.isnull().sum()
print(dado_nulo)
df.dtypes
```

```python
df['dias_envio']=(df['Ship Date']-df['Order Date']).dt.days
df[['Order Date', 'Ship Date', 'dias_envio']].head()
```

```python
analise_envio=df.groupby('dias_envio')['Profit'].agg(['count', 'sum', 'mean'])
print(analise_envio)
"""Nesta primeira análise, o prazo de envio não apresenta relação direta com prejuízo. 
Todos os prazos entre 0 e 7 dias tiveram lucro total e lucro médio positivos. 
O prazo de 4 dias concentra mais vendas, mas isso parece refletir volume operacional, não necessariamente maior rentabilidade."""
```

```python
calculo_margem=df['Profit']/df['Sales']
df['margem_lucro']=calculo_margem
df[['Sub-Category','Sales','Profit','margem_lucro']].head()
```

```python
margem_subcategoria=df.groupby('Sub-Category')['margem_lucro'].mean().sort_values()
print(margem_subcategoria)
#Na análise de margem média por venda, Binders, Appliances, Tables e Bookcases apresentam margem média negativa. 
#Porém, na análise de lucro total, Tables continua sendo o maior detrator financeiro.
```

```python
resumo_subcategoria=df.groupby('Sub-Category').agg(
    total_vendas=('Sales','count'),
    lucro_total=('Profit','sum'),
    margem_media=('margem_lucro','mean')
).sort_values('lucro_total')
print(resumo_subcategoria)
```

```python
#analise binders, pois foi o que mais teve margem media negativa, mas teve lucro total positivo
binders=df[df['Sub-Category']=='Binders']
binders_prejuizo=binders[binders['Profit']<0]
binders_lucro=binders[binders['Profit']>=0]

#quantidade de vendas
qt_vendas_prejuizo=len(binders_prejuizo)
qt_vendas_lucro=len(binders_lucro)

#vendas totais
vendas_totais_prejuizo=binders_prejuizo['Sales'].sum()
vendas_totais_lucro=binders_lucro['Sales'].sum()

#vendas media
vendas_media_prejuizo=binders_prejuizo['Sales'].mean()
vendas_media_lucro=binders_lucro['Sales'].mean()

#lucro total
lucro_total_prejuizo=binders_prejuizo['Profit'].sum()
lucro_total_lucro=binders_lucro['Profit'].sum()

#lucro medio
lucro_medio_prejuizo=binders_prejuizo['Profit'].mean()
lucro_medio_lucro=binders_lucro['Profit'].mean()

#margem media
margem_media_lucro=binders_lucro['margem_lucro'].mean()
margem_media_prejuizo=binders_prejuizo['margem_lucro'].mean()

#desconto medio
desc_medio_prejuizo=binders_prejuizo['Discount'].mean()
desc_medio_lucro=binders_lucro['Discount'].mean()

comparacao_binders=pd.DataFrame({
    'grupo': ['Binders com prejuizo', 'Binders com lucro'],
    'quantidade_vendas': [qt_vendas_prejuizo, qt_vendas_lucro],
    'vendas_totais': [vendas_totais_prejuizo, vendas_totais_lucro],
    'vendas_media': [vendas_media_prejuizo, vendas_media_lucro],
    'lucro_total': [lucro_total_prejuizo, lucro_total_lucro],
    'margem_media': [margem_media_prejuizo, margem_media_lucro],
    'desconto_medio': [desc_medio_prejuizo, desc_medio_lucro]
})
comparacao_binders=comparacao_binders.round(2)
print(comparacao_binders)
```

```python
#analise desconto binders
binders_high_discount=binders[binders['Discount'] >= 0.2]
binders_low_discount=binders[binders['Discount'] < 0.2]

print(binders_high_discount)
print(binders_low_discount)
```

```python
#analise de lucro dos descontos
lucro_high_discount=binders_high_discount['Profit'].sum()
print('Lucro de desconto >=20%:', lucro_high_discount)

lucro_low_discount=binders_low_discount['Profit'].sum()
print('Lucro de desconto <20%:', lucro_low_discount)
```

```python
#analise de margem media de desconto
margem_high_discount=binders_high_discount['margem_lucro'].mean()
print('margem de lucro com desconto alto:', margem_high_discount)

margem_low_discount=binders_low_discount['margem_lucro'].mean()
print('margem de lucro com desconto baixo', margem_low_discount)
```

```python
#analise de tables por prejuizo e lucro Profit
tables=df[df['Sub-Category'] == 'Tables']
tables_prejuizo=tables[tables['Profit']<0]
tables_lucro=tables[tables['Profit']>=0]

#quantidade de vendas
qt_vendas_prejuizo_tables=len(tables_prejuizo)
qt_vendas_lucro_tables=len(tables_lucro)

#vendas totais
vendas_totais_prejuizo_tables=tables_prejuizo['Sales'].sum()
vendas_totais_lucro_tables=tables_lucro['Sales'].sum()

#valor medio de venda
vendas_media_prejuizo_tables=tables_prejuizo['Sales'].mean()
vendas_media_lucro_tables=tables_lucro['Sales'].mean()

#lucro total
lucro_total_prejuizo_tables=tables_prejuizo['Profit'].sum()
lucro_total_lucro_tables=tables_lucro['Profit'].sum()

#margem media
margem_media_prejuizo_tables=tables_prejuizo['margem_lucro'].mean()
margem_media_lucro_tables=tables_lucro['margem_lucro'].mean()

#desconto medio
desc_medio_prejuizo_tables=tables_prejuizo['Discount'].mean()
desc_medio_lucro_tables=tables_lucro['Discount'].mean()

comparacao_tables=pd.DataFrame({
    'grupo':['tables com prejuizo', 'tables com lucro'],
    'quantidade_vendas':[qt_vendas_prejuizo_tables, qt_vendas_lucro_tables],
    'vendas_totais':[vendas_totais_prejuizo_tables, vendas_totais_lucro_tables],
    'vendas_media':[vendas_media_prejuizo_tables, vendas_media_lucro_tables],
    'lucro_total':[lucro_total_prejuizo_tables, lucro_total_lucro_tables],
    'margem_media':[margem_media_prejuizo_tables, margem_media_lucro_tables],
    'desconto_medio':[desc_medio_prejuizo_tables, desc_medio_lucro_tables]
})
comparacao_tables=comparacao_tables.round(2)
comparacao_tables
```

```python
comparacao_tables_binders=pd.concat(
    [comparacao_tables, comparacao_binders],
    ignore_index=True
)

comparacao_tables_binders.sort_values('lucro_total')
#Binders com prejuízo perde mais que Tables com prejuízo, mas ela não mostra claramente que Binders com lucro compensa isso. 
#Ou seja tables da um estrago maior no final
```

```python
comparacao_final_subcategoria=pd.DataFrame({
    'subcategoria':['Tables','Binders'],
    'lucro_grupo_prejuizo':[lucro_total_prejuizo_tables, lucro_total_prejuizo],
    'lucro_grupo_lucro':[lucro_total_lucro_tables, lucro_total_lucro],
    'lucro_total_subcategoria':[tables['Profit'].sum(), binders['Profit'].sum()],
    'margem_media_subcategoria':[tables['margem_lucro'].mean(), binders['margem_lucro'].mean()],
    'desconto_medio_prejuizo':[desc_medio_prejuizo_tables, desc_medio_prejuizo],
    'status_final':['fecha com prejuizo', 'fecha com lucro']
    
})
comparacao_final_subcategoria.round(2)
```
