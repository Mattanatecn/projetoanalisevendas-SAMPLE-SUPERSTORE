```python
# Importa a biblioteca Pandas, essencial para manipulação e análise de tabelas de dados
import pandas as pd
```

```python
# Carrega o conjunto de dados bruto. 
# encoding='latin1' é usado porque o arquivo possui caracteres especiais (acentos, etc) que o padrão UTF-8 não reconhece
df = pd.read_csv('../dados/bruto/Sample - Superstore.csv', encoding='latin1')
```

```python
# Exibe as primeiras 5 linhas do DataFrame para ter uma visão geral das colunas e dos dados
df.head()
```

```python
# Analisa a estrutura do DataFrame: tipos de dados, se há valores nulos e o volume de memória usado
df.info()
```

```python
# Verifica a existência de linhas idênticas para garantir que não haja contagem duplicada de vendas
print(f"Linhas duplicadas: {df.duplicated().sum()}")
```

```python
# Cálculo do KPI de Vendas Totais: soma de todo o faturamento da empresa
vendas=df['Sales'].sum()
print(vendas)
```

```python
# Cálculo do KPI de Lucro Total: soma de todos os lucros (positivos e negativos)
lucro_total=df['Profit'].sum()
print(lucro_total)
```

```python
# Identifica as vendas que geraram prejuízo (lucro < 0) e soma esse prejuízo total
# Isso ajuda a entender a escala do problema financeiro da empresa
vendas_prejuizo=df[df['Profit'] < 0]
lucro_negativo=vendas_prejuizo['Profit'].sum()
print(lucro_negativo)
```

```python
# Agrupa os dados por Categoria para descobrir qual setor do negócio é mais lucrativo
lucro_categoria=df.groupby('Category')['Profit'].sum()
print(lucro_categoria)
```

```python
# Detalha o lucro por Subcategoria para localizar exatamente onde o dinheiro está sendo perdido
lucro_subcategorias=df.groupby('Sub-Category')['Profit'].sum()
print(lucro_subcategorias)
```

```python
# Ordena os lucros das subcategorias do menor para o maior
# Isso evidencia rapidamente a subcategoria 'Tables' como a principal fonte de prejuízo
lucro_ordenado=lucro_subcategorias.sort_values()
print(lucro_ordenado)
```

```python
# PASSO 4: INVESTIGAÇÃO DA CAUSA RAIZ
# Isolamos apenas as vendas de 'Tables' para analisar o comportamento específico desta subcategoria
tables=df[df['Sub-Category']=='Tables']
print(tables)
```

```python
# Filtrando vendas de mesas com descontos agressivos (20% ou mais)
# Hipótese: Descontos altos estão anulando a margem de lucro
df_high_discount=tables[tables['Discount']>=0.2]
print(df_high_discount)
```

```python
# Filtrando vendas de mesas com descontos baixos (menos de 20%)
# Serve como grupo de controle para comparar se o prejuízo persiste sem descontos altos
df_low_discount=tables[tables['Discount']<0.2]
print(df_low_discount)
```

```python
# Calcula o lucro total para o grupo de descontos altos
# Resultado esperado: Lucro fortemente negativo
lucro_negativo=df_high_discount['Profit'].sum()
print(f"Lucro Total (Descontos >= 20%): {lucro_negativo}")
```

```python
# Calcula o lucro total para o grupo de descontos baixos
# Resultado esperado: Lucro positivo ou prejuízo irrelevante
lucro_positivo=df_low_discount['Profit'].sum()
print(f"Lucro Total (Descontos < 20%): {lucro_positivo}")
```

```python
# LIMPEZA DE DADOS: Transformando colunas de data de texto (str) para formato datetime
# Isso permite realizar cálculos de tempo, sazonalidade e análises temporais
df['Order Date']=pd.to_datetime(df['Order Date'])
df['Ship Date']=pd.to_datetime(df['Ship Date'])
df.info()
```

```python
dado_nulo=df.isnull().sum()
print(dado_nulo)
df.dtypes
```

```python
df['dias_envio']=(df['Ship Date']-df['Order Date']).dt.days
df[['Order Date', 'Ship Date', 'dias_envio']].head()
```

```python
analise_envio=df.groupby('dias_envio')['Profit'].agg(['count', 'sum', 'mean'])
print(analise_envio)
"""Nesta primeira análise, o prazo de envio não apresenta relação direta com prejuízo. 
Todos os prazos entre 0 e 7 dias tiveram lucro total e lucro médio positivos. 
O prazo de 4 dias concentra mais vendas, mas isso parece refletir volume operacional, não necessariamente maior rentabilidade."""
```

```python
calculo_margem=df['Profit']/df['Sales']
df['margem_lucro']=calculo_margem
df[['Sub-Category','Sales','Profit','margem_lucro']].head()
```

```python
margem_subcategoria=df.groupby('Sub-Category')['margem_lucro'].mean().sort_values()
print(margem_subcategoria)
#Na análise de margem média por venda, Binders, Appliances, Tables e Bookcases apresentam margem média negativa. 
#Porém, na análise de lucro total, Tables continua sendo o maior detrator financeiro.
```

```python
resumo_subcategoria=df.groupby('Sub-Category').agg(
    total_vendas=('Sales','count'),
    lucro_total=('Profit','sum'),
    margem_media=('margem_lucro','mean')
).sort_values('lucro_total')
print(resumo_subcategoria)
```
